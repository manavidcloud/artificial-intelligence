```python
from langchain_core.documents import Document
# Create individual documents
doc1 = Document(
    page_content="LangChain is a framework for developing applications powered by language models.",
    metadata={"source": "introduction.pdf", "page": 1}
)
doc2 = Document(
    page_content="It enables applications to connect a model to other sources of data.",
    metadata={"source": "introduction.pdf", "page": 2}
)
doc3 = Document(
    page_content="Documents are the core unit of data in LangChain.",
    metadata={"source": "api_reference.txt", "author": "DevTeam"}
)
# Create the list
my_documents = [doc1, doc2, doc3]
# Verify the list
print(f"Total documents: {len(my_documents)}")
print(f"Type of first doc: {type(my_documents[0])}")
print("\n++++++++++++++++++++++++++++++++++++++++++++++++\n")
print(f"Content of first doc: {my_documents[0].page_content}")
print(f"Type of page_content: {type(my_documents[0].page_content)}")
print("\n++++++++++++++++++++++++++++++++++++++++++++++++\n")
print(f"Metadata of first doc: {my_documents[0].metadata}")
print(f"Type of metadata: {type(my_documents[0].metadata)}")
print("\n++++++++++++++++++++++++++++++++++++++++++++++++\n")
print(f"Source of first doc: {my_documents[0].metadata['source']}")
print(f"Type of source: {type(my_documents[0].metadata['source'])}")
print("\n++++++++++++++++++++++++++++++++++++++++++++++++\n")
print(f"Author of first doc: {my_documents[0].metadata.get('author', 'N/A')}")  
print(f"Type of author: {type(my_documents[0].metadata.get('author', 'N/A'))}") 
print("\n++++++++++++++++++++++++++++++++++++++++++++++++\n")
print(f"Author of third doc: {my_documents[2].metadata.get('author', 'N/A')}")  
print(f"Type of author: {type(my_documents[2].metadata.get('author', 'N/A'))}") 
```

```python
from langchain_community.document_loaders import TextLoader
loader=TextLoader('cricket.txt',encoding='utf-8') ## making the connection to the file
docs=loader.load() 
print(f'The type of docs is: {type(docs)}')
print(f'The number of docs is: {len(docs)}')
print(f'The type of the first doc is: {type(docs[0])}')
```

```python
from langchain_openai import ChatOpenAI
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import PromptTemplate
from dotenv import load_dotenv,find_dotenv
load_dotenv(find_dotenv())

llm=ChatOpenAI(model_name='gpt-5-nano')
prompt = PromptTemplate(
    template='Write a summary for the following poem - \n {poem}',
    input_variables=['poem']
)
output_parser=StrOutputParser()

chain=prompt|llm|output_parser
result=chain.invoke({'poem':docs[0].page_content})
print(result) 
```
