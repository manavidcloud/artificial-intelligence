# 3. Our First Neural Network

We will build:

- 5 input features
- 1 output neuron
- Sigmoid activation

Architecture:

Input(5) → Linear → Sigmoid → Output(1)

## Equation

$$
z = Wx + b
$$

$$
\hat y = sigmoid(z)
$$


````python
import torch
import torch.nn as nn

class Model(nn.Module):

    def __init__(self, num_features):

        super().__init__()

        self.linear = nn.Linear(num_features, 1)
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):

        out = self.linear(x)
        out = self.sigmoid(out)

        return out
````

# 4. Understanding `super().__init__()` and our class Model

Our class inherits from:

```python
nn.Module
```

Therefore:

```python
super().__init__()
```

calls the constructor of the parent class.

Without this line, PyTorch cannot properly register layers and parameters.


````python
# No execution needed.
# This is a theory section.
````

Here's how the `Model` class reflects the architecture and mathematical idea:

*   **`__init__(self, num_features)`**: This method is the constructor where we *define* the layers (building blocks) of our neural network. It's like setting up the blueprint.
    *   **`super().__init__()`**: Calls the constructor of the parent class `nn.Module` to ensure proper initialization.
    *   **`self.linear = nn.Linear(num_features, 1)`**: This creates an instance of a linear layer. At this stage, you are *initializing* the layer, which includes creating its learnable parameters (weights `W` and biases `b`). The calculation `z = Wx + b` *does not happen yet*.
    *   **`self.sigmoid = nn.Sigmoid()`**: This creates an instance of the sigmoid activation function.

*   **`forward(self, x)`**: This method defines the forward pass, i.e., how the input `x` flows through the network to produce an output. This is where the *actual calculations* happen.
    *   **`out = self.linear(x)`**: **This is where the calculation `z = Wx + b` actually occurs!** When you call `self.linear(x)`, you are passing the input `x` through the `linear` layer's internal `forward` method, which performs the matrix multiplication of `x` with `W` and adds the bias `b` to produce `z`.
    *   **`out = self.sigmoid(out)`**: The result `z` from the linear layer is then passed through the `sigmoid` activation function. This step corresponds to `y_pred = sigmoid(z)`.
    *   **`return out`**: The final predicted output `y_pred` is returned.

## Forward Pass: What Happens Inside `model(features)`?

When we execute:

```python
predictions = model(features)
```

PyTorch automatically calls:

```python
model.forward(features)
```

Inside the model, the first operation is:

```python
z = self.linear(features)
```

Mathematically:

$$
Z = XW^T + b
$$

where

$$
X =
\begin{bmatrix}
x_{11} & x_{12} & x_{13} & x_{14} & x_{15} \\
x_{21} & x_{22} & x_{23} & x_{24} & x_{25} \\
\vdots & \vdots & \vdots & \vdots & \vdots \\
x_{10,1} & x_{10,2} & x_{10,3} & x_{10,4} & x_{10,5}
\end{bmatrix}
$$

Shape:

$$
(10 \times 5)
$$

The weight matrix learned by the model is:

$$
W =
\begin{bmatrix}
w_1 & w_2 & w_3 & w_4 & w_5
\end{bmatrix}
$$

Shape:

$$
(1 \times 5)
$$

PyTorch internally uses:

$$
W^T =
\begin{bmatrix}
w_1\\
w_2\\
w_3\\
w_4\\
w_5
\end{bmatrix}
$$

Shape:

$$
(5 \times 1)
$$

Therefore:

$$
(10 \times 5)
\times
(5 \times 1)
=
(10 \times 1)
$$

The matrix multiplication becomes:

$$
\begin{bmatrix}
x_{11} & x_{12} & x_{13} & x_{14} & x_{15}\\
x_{21} & x_{22} & x_{23} & x_{24} & x_{25}\\
\vdots & \vdots & \vdots & \vdots & \vdots\\
x_{10,1} & x_{10,2} & x_{10,3} & x_{10,4} & x_{10,5}
\end{bmatrix}
\times
\begin{bmatrix}
w_1\\
w_2\\
w_3\\
w_4\\
w_5
\end{bmatrix}
+
b
$$

Result:

$$
Z =
\begin{bmatrix}
z_1\\
z_2\\
\vdots\\
z_{10}
\end{bmatrix}
$$

Shape:

$$
(10 \times 1)
$$

Each output is calculated as:

$$
z_i =
w_1x_{i1}
+
w_2x_{i2}
+
w_3x_{i3}
+
w_4x_{i4}
+
w_5x_{i5}
+
b
$$

Next, the Sigmoid activation function is applied:

$$
\hat y =
\sigma(z)
=
\frac{1}{1+e^{-z}}
$$

Therefore:

$$
Predictions =
\sigma(XW^T+b)
$$

Final output shape:

$$
(10 \times 1)
$$

This is exactly why:

```python
print(predictions.shape)
```

produces:

```python
torch.Size([10,1])
```

# 9. Multi-Layer Neural Network

Now we build:

Input(5)
→ Hidden Layer(3 neurons)
→ ReLU
→ Output Layer(1 neuron)
→ Sigmoid

## Parameters

First Layer:

$$
5 \times 3 = 15
$$

Second Layer:

$$
3 \times 1 = 3
$$

Total weights:

$$
18
$$

Biases:

$$
3 + 1 = 4
$$


````python
class DeepModel(nn.Module):

    def __init__(self, num_features):

        super().__init__()

        self.linear1 = nn.Linear(num_features, 3)
        self.relu = nn.ReLU()

        self.linear2 = nn.Linear(3, 1)
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):

        out = self.linear1(x)
        out = self.relu(out)

        out = self.linear2(out)
        out = self.sigmoid(out)

        return out
````

# 10. Test the Multi-Layer Model

# 11. Why `nn.Sequential`?

When networks become larger, writing:

```python
out = layer1(x)
out = relu(out)
out = layer2(out)
...
```

becomes repetitive.

`nn.Sequential` automatically chains layers together.


````python
class SequentialModel(nn.Module):

    def __init__(self, num_features):

        super().__init__()

        self.network1 = nn.Sequential(
            nn.Linear(num_features, 3),
            nn.ReLU(),
            nn.Linear(3, 1),
            nn.Sigmoid()
        )

    def forward(self, x):
        return self.network1(x)
````

# 12. Run the Sequential Model


````python
seq_model = SequentialModel(5)

preds = seq_model(features)

print(preds.shape)
preds[:5]
````

````python
from torchinfo import summary

summary(seq_model, input_size=(1,5))
````

# 13. Binary Classification Loss

For binary classification we often use:

```python
nn.BCELoss()
```

Mathematical Formula:

$$
Loss = -[y\log(\hat y)+(1-y)\log(1-\hat y)]
$$

Lower loss means better predictions.


````python
loss_function = nn.BCELoss()

print(loss_function)
````

# 14. Create Target Labels


````python
y_train = torch.randint(0, 2, (10,1)).float()

print(y_train)
````

# 15. Calculate Loss


````python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Ensure the model is on the correct device
seq_model = seq_model.to(device)

# Ensure the input features are on the correct device
features = features.to(device)

# Ensure the target labels are on the correct device
y_train = y_train.to(device)

preds = seq_model(features)

loss = loss_function(preds, y_train)

print("Loss =", loss.item())
````

````python
preds.shape , y_train.shape
````

# 16. Backpropagation

PyTorch calculates gradients automatically.

Steps:

1. Forward Pass
2. Compute Loss
3. Backward Pass

```python
loss.backward()
```


````python
loss.backward()

print("Gradient of first layer weights")
print(seq_model.network1[0].weight.grad)

print("\nGradient of first layer bias")
print(seq_model.network1[0].bias.grad)

print("$"*50)

print("\nGradient of second layer weights")
print(seq_model.network1[2].weight.grad)

print("\nGradient of second layer bias")
print(seq_model.network1[2].bias.grad)
````

# 17. Optimizers

Instead of manually updating:

$$
w = w - \eta \frac{\partial L}{\partial w}
$$

we use an optimizer.

Most common:

```python
torch.optim.SGD
```


````python
optimizer = torch.optim.SGD(seq_model.parameters(),lr=0.01)

print(optimizer)
````

# 18. Understanding `model.parameters()`

This function returns all trainable parameters:

- Weights

````python
for parameter in seq_model.parameters():
    print(parameter.shape)
````

# 19. Full Training Loop

Standard PyTorch Training Flow:

```text
Zero Gradients
      ↓
Forward Pass
      ↓
Loss Calculation
      ↓
Backward Pass
      ↓
Optimizer Step
```


````python
for epoch in range(5):

    optimizer.zero_grad()

    predictions = seq_model(features)

    loss = loss_function(predictions, y_train)

    loss.backward()

    optimizer.step()

    print(f"Epoch {epoch+1}: Loss = {loss.item():.4f}")
````
