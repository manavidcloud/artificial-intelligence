### 3. Data Preprocessing

- **Dropping Irrelevant Columns**: The 'id' column is dropped as it's just an identifier and 'Unnamed: 32' appears to be an empty or irrelevant column from the data preview.

- **Label Encoding**: The target variable `diagnosis` is categorical ('M' for Malignant, 'B' for Benign). `LabelEncoder` converts these categorical labels into numerical format (e.g., 0 and 1), which is required for model training.

### 4. Converting to PyTorch Tensors

Before feeding the data into a PyTorch model, the preprocessed NumPy arrays (`X_train`, `X_test`, `y_train`, `y_test`) are converted into PyTorch tensors. This is done using `torch.from_numpy()`.

```python
X_train_tensor = torch.from_numpy(X_train)
X_test_tensor = torch.from_numpy(X_test)
y_train_tensor = torch.from_numpy(y_train)
y_test_tensor = torch.from_numpy(y_test)
```

### 5. Defining the Neural Network Model

This section defines a simple neural network class `MySimpleNN`. It's a basic logistic regression model, which is a single-layer neural network.

- **`__init__`**: Initializes the `weights` and `bias` of the model. The weights are randomly initialized, and the bias is initialized to zeros. `requires_grad=True` is set to enable automatic gradient calculation by PyTorch.
- **`forward`**: This method defines the forward pass of the network. It calculates the linear combination of inputs and weights, adds the bias, and then applies a sigmoid activation function to produce predictions (`y_pred`). The sigmoid function squashes the output between 0 and 1, making it suitable for binary classification.
- **`loss_function`**: Implements the binary cross-entropy loss function, which is commonly used for binary classification tasks. It measures the dissimilarity between the predicted probabilities and the true labels. A small `epsilon` is added to `y_pred` to prevent `log(0)` errors.

```python
class MySimpleNN():

  def __init__(self, X):

    self.weights = torch.rand(X.shape[1], 1, dtype=torch.float64, requires_grad=True)
    self.bias = torch.zeros(1, dtype=torch.float64, requires_grad=True)

  def forward(self, X):
    z = torch.matmul(X, self.weights) + self.bias
    y_pred = torch.sigmoid(z)
    return y_pred

  def loss_function(self, y_pred, y):
    # Clamp predictions to avoid log(0)
    epsilon = 1e-7
    y_pred = torch.clamp(y_pred, epsilon, 1 - epsilon)

    # Calculate loss
    loss = -(y_train_tensor * torch.log(y_pred) + (1 - y_train_tensor) * torch.log(1 - y_pred)).mean()
    return loss



```

### 6. Important Parameters

This cell defines key hyperparameters for the training process:

- **`learning_rate`**: Controls the step size during parameter updates. A smaller learning rate means smaller updates, while a larger one means bigger updates.
- **`epochs`**: Represents the number of times the entire training dataset is passed through the neural network. More epochs can lead to better learning but also risk overfitting.

```python
learning_rate = 0.1
epochs = 25
```

### 7. Training Pipeline

This is the core training loop:

- **Model Instantiation**: An instance of `MySimpleNN` is created.
- **Epoch Loop**: The training runs for a specified number of `epochs`.
  - **Forward Pass**: `model.forward(X_train_tensor)` calculates the predictions for the training data.
  - **Loss Calculation**: `model.loss_function(y_pred, y_train_tensor)` computes the loss based on the predictions and actual labels.
  - **Backward Pass**: `loss.backward()` calculates the gradients of the loss with respect to the model's weights and bias. This is where PyTorch's automatic differentiation shines.
  - **Parameter Update**: The `weights` and `bias` are updated using the calculated gradients and the `learning_rate`. This step uses `torch.no_grad()` to ensure that these updates are not part of the computational graph for gradient calculation.
  - **Zero Gradients**: `model.weights.grad.zero_()` and `model.bias.grad.zero_()` reset the gradients to zero after each update to prevent accumulation from previous iterations.
  - **Loss Reporting**: The loss is printed for each epoch to monitor the training progress.

```python
# Hyper Parmeter
learning_rate = 0.1
epochs = 25
# create model
model = MySimpleNN(X_train_tensor)

# define loop
for epoch in range(epochs):

  # forward pass
  y_pred = model.forward(X_train_tensor)

  # loss calculate
  loss = model.loss_function(y_pred, y_train_tensor)

  # backward pass
  loss.backward()

  # parameters update
  with torch.no_grad():
    model.weights -= learning_rate * model.weights.grad
    model.bias -= learning_rate * model.bias.grad

  # zero gradients
  model.weights.grad.zero_()
  model.bias.grad.zero_()

  # print loss in each epoch
  print(f'Epoch: {epoch + 1}, Loss: {loss.item()}')
```

### 8. Model Evaluation

After training, the model's performance is evaluated on the unseen test data:

- **`torch.no_grad()`**: This context manager ensures that no gradients are computed during evaluation, saving memory and computation.
- **Prediction**: `model.forward(X_test_tensor)` makes predictions on the test set.
- **Thresholding**: The continuous probability predictions from the sigmoid function are converted into binary class labels (0 or 1) by applying a threshold (here, 0.9). If `y_pred` is greater than 0.9, it's classified as 1 (Malignant); otherwise, it's 0 (Benign).
- **Accuracy Calculation**: The `accuracy` is calculated by comparing the predicted labels with the true test labels and finding the mean of correct predictions.

```python
# model evaluation
with torch.no_grad():
  y_pred = model.forward(X_test_tensor)
  y_pred = (y_pred > 0.9).float()
  accuracy = (y_pred == y_test_tensor).float().mean()
  print(f'Accuracy: {accuracy.item()}')
```
