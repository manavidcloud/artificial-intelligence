## 2. Creating Tensors

Tensors are the fundamental data structure in PyTorch, similar to NumPy arrays. They can be created in various ways, initialized with specific values, or generated randomly.



### Initializing Tensors

PyTorch provides in built functions to create tensors with uninitialized values, zeros, ones, or random numbers.

```python
# torch.empty(shape): Creates a tensor with uninitialized data. The values will be whatever was in memory at the time.
print("Empty tensor:\n", torch.empty(2,3))

# torch.zeros(shape): Creates a tensor filled with zeros.
print("\nZeros tensor:\n", torch.zeros(2,3))

# torch.ones(shape): Creates a tensor filled with ones.
print("\nOnes tensor:\n", torch.ones(2,3))

# torch.rand(shape): Creates a tensor with random numbers from a uniform distribution between 0 and 1.
print("\nRandom tensor:\n", torch.rand(2,3))
```

To ensure reproducibility of random tensors, you can use `torch.manual_seed()`. This sets the seed for the CPU, making random number generation deterministic.

```python
# Using torch.manual_seed for reproducibility
torch.manual_seed(100)
print("Random tensor with seed 100:\n", torch.rand(2,3))

# Calling rand again without resetting the seed will produce new random numbers
print("\nAnother random tensor without resetting seed:\n", torch.rand(2,3))

# Resetting the seed to the same value will produce the same sequence of random numbers again
torch.manual_seed(100)
print("\nRandom tensor after resetting seed to 100:\n", torch.rand(2,3))
```

### Creating Tensors from Data

You can create tensors directly from Python lists or NumPy arrays using `torch.tensor()`.

```python
# Create a tensor from a Python list
my_list = [[1,2,3],[4,5,6]]
print("Tensor from list:\n", torch.tensor(my_list))
```

### Specialized Tensor Creation Functions

PyTorch offers several functions for creating tensors with specific patterns, like sequences, identity matrices, or tensors filled with a constant value.

```python
# torch.arange(start, end, step): Returns a 1-D tensor of size `(end-start)/step` with values from `start` to `end` with `step`.
print("Using arange (0 to 10, step 2):", torch.arange(0,10,2))

# torch.linspace(start, end, steps): Returns a 1-D tensor of `steps` equally spaced points between `start` and `end`.
print("Using linspace (0 to 10, 10 steps):", torch.linspace(0,10,10))

# torch.eye(n): Returns a 2-D tensor with ones on the diagonal and zeros elsewhere (identity matrix).
print("Using eye (5x5 identity matrix):\n", torch.eye(5))

# torch.full(shape, fill_value): Creates a tensor of `shape` filled with `fill_value`.
print("Using full (3x3, filled with 5):\n", torch.full((3, 3), 5))
```

## 3. Tensor Shapes

The shape of a tensor describes its dimensions. Understanding and manipulating tensor shapes is crucial for deep learning models.



### Inspecting Shape

You can get the shape of a tensor using the `.shape` attribute.

```python
x = torch.tensor([[1,2,3],[4,5,6]])
print("Original tensor x:\n", x)
print("Shape of x:", x.shape)
```

### Creating Tensors with Similar Shapes

PyTorch allows creating new tensors with the same shape as an existing tensor, often useful for creating masks or temporary storage.

```python
# torch.empty_like(input): Returns an uninitialized tensor with the same size as `input`.
print("Empty like x:\n", torch.empty_like(x))

# torch.zeros_like(input): Returns a tensor filled with zeros, with the same size as `input`.
print("\nZeros like x:\n", torch.zeros_like(x))

# torch.ones_like(input): Returns a tensor filled with ones, with the same size as `input`.
print("\nOnes like x:\n", torch.ones_like(x))

# torch.rand_like(input, dtype): Returns a tensor with random numbers from a uniform distribution, with the same size as `input`.
# You can also specify the dtype, e.g., torch.float32 for floating-point random numbers.
print("\nRandom like x (float32):\n", torch.rand_like(x, dtype=torch.float32))
```

## 4. Tensor Data Types

Each tensor has a specific data type (`dtype`) that determines the kind of values it can store (e.g., integers, floating-point numbers, booleans) and how much memory it consumes. Choosing the correct data type is important for memory efficiency and numerical precision.



### Checking and Assigning Data Types

You can check a tensor's data type using `.dtype` and assign a specific data type during creation or conversion.

```python
x = torch.tensor([[1,2,3],[4,5,6]])
print("Data type of x:", x.dtype)

# Assign data type during tensor creation
print("\nTensor with int32 dtype:", torch.tensor([1.0,2.0,3.0], dtype=torch.int32))
print("Tensor with float64 dtype:", torch.tensor([1,2,3], dtype=torch.float64))

# Convert a tensor to a different data type using .to()
print("\nTensor x converted to float32:\n", x.to(torch.float32))
```

## 5. Mathematical Operations

PyTorch tensors support a wide range of mathematical operations, which can be categorized into scalar, element-wise, reduction, matrix, comparison, and special functions.

### Scalar Operations

Scalar operations involve performing an operation between a tensor and a single number (scalar). The operation is applied to every element of the tensor.

```python
x = torch.rand(2,2)
print("Original tensor x:\n", x)

print("\nAddition (x + 2):\n", x + 2)
print("Subtraction (x - 2):\n", x - 2)
print("Multiplication (x * 3):\n", x * 3)
print("Division (x / 3):\n", x / 3)
print("Integer Division ((x * 100) // 3):\n", (x * 100) // 3)
print("Modulo (((x * 100) // 3) % 2):\n", ((x * 100) // 3) % 2)
print("Power (x ** 2):\n", x**2)
```

### Element-Wise Operations

Element-wise operations apply a function or operation to each corresponding element of two tensors of the same shape.

```python
a = torch.rand(2,3)
b = torch.rand(2,3)

print("Tensor a:\n", a)
print("Tensor b:\n", b)

print("\nElement-wise addition (a + b):\n", a + b)
print("Element-wise subtraction (a - b):\n", a - b)
print("Element-wise multiplication (a * b):\n", a * b)
print("Element-wise division (a / b):\n", a / b)
print("Element-wise power (a ** b):\n", a ** b)
print("Element-wise modulo (a % b):\n", a % b)
```

Other common element-wise functions include absolute value, negation, rounding, ceiling, floor, and clamping.

```python
c = torch.tensor([1, -2, 3, -4])
print("Tensor c:\n", c)
print("Absolute value (torch.abs(c)):", torch.abs(c))
print("Negation (torch.neg(c)):", torch.neg(c))

d = torch.tensor([1.9, 2.3, 3.7, 4.4])
print("\nTensor d:\n", d)
print("Round (torch.round(d)):", torch.round(d))
print("Ceil (torch.ceil(d)):", torch.ceil(d))
print("Floor (torch.floor(d)):", torch.floor(d))
print("Clamp (torch.clamp(d, min=2, max=3)):", torch.clamp(d, min=2, max=3))
```

### Reduction Operations

Reduction operations reduce a tensor along one or more dimensions, resulting in a tensor with fewer dimensions (or a scalar). Common reductions include sum, mean, median, max, min, product, standard deviation, and variance.

```python
e = torch.randint(low=0, high=10, size=(2,3), dtype=torch.float32)
print("Tensor e:\n", e)

print("\nSum of all elements (torch.sum(e)):", torch.sum(e))
print("Sum along columns (dim=0):", torch.sum(e, dim=0))
print("Sum along rows (dim=1):", torch.sum(e, dim=1))

print("\nMean of all elements (torch.mean(e)):", torch.mean(e))
print("Mean along columns (dim=0):", torch.mean(e, dim=0))

print("\nMedian of all elements (torch.median(e)):", torch.median(e))

print("\nMaximum of all elements (torch.max(e)):", torch.max(e))
print("Minimum of all elements (torch.min(e)):", torch.min(e))

print("\nProduct of all elements (torch.prod(e)):", torch.prod(e))

print("\nStandard deviation of all elements (torch.std(e)):", torch.std(e))

print("\nVariance of all elements (torch.var(e)):", torch.var(e))

# argmax/argmin return the index of the maximum/minimum value
print("\nIndex of maximum value (torch.argmax(e)):", torch.argmax(e))
print("Index of minimum value (torch.argmin(e)):", torch.argmin(e))
```

### Matrix Operations

PyTorch supports standard linear algebra operations like matrix multiplication, dot product, transpose, determinant, and inverse.

```python
f = torch.randint(low=0, high=10, size=(2,3))
g = torch.randint(low=0, high=10, size=(3,2))

print("Tensor f (2x3):\n", f)
print("Tensor g (3x2):\n", g)

# Matrix multiplication (f * g) - the inner dimensions must match (3 in this case)
print("\nMatrix multiplication (torch.matmul(f, g)):\n", torch.matmul(f, g))

vector1 = torch.tensor([1, 2])
vector2 = torch.tensor([3, 4])
print("\nVector 1:", vector1)
print("Vector 2:", vector2)
# Dot product of two 1-D tensors
print("Dot product (torch.dot(vector1, vector2)):", torch.dot(vector1, vector2))

# Transpose: Swaps the dimensions of a tensor
# For a 2D tensor, it swaps rows and columns.
print("\nTranspose of f (torch.transpose(f, 0, 1)):\n", torch.transpose(f, 0, 1))

h = torch.randint(low=0, high=10, size=(3,3), dtype=torch.float32)
print("\nTensor h (3x3):\n", h)

# Determinant of a square matrix
print("Determinant of h (torch.det(h)):", torch.det(h))

# Inverse of a square matrix
# Note: The matrix must be invertible (non-singular).
print("Inverse of h (torch.inverse(h)):\n", torch.inverse(h))
```

### Comparison Operations

Comparison operations perform element-wise comparisons between two tensors or a tensor and a scalar, returning a boolean tensor.

```python
i = torch.randint(low=0, high=10, size=(2,3))
j = torch.randint(low=0, high=10, size=(2,3))

print("Tensor i:\n", i)
print("Tensor j:\n", j)

print("\nGreater than (i > j):\n", i > j)
print("Less than (i < j):\n", i < j)
print("Equal to (i == j):\n", i == j)
print("Not equal to (i != j):\n", i != j)
# Add more comparison operations as needed, e.g., >=, <=
```

### Special Functions

PyTorch also includes various special mathematical functions commonly used in deep learning, such as logarithmic, exponential, square root, sigmoid, softmax, and ReLU.

```python
k = torch.randint(low=0, high=10, size=(2,3), dtype=torch.float32)
print("Tensor k:\n", k)

# Natural logarithm
# Note: log(0) is -inf, so input tensor should ideally contain positive values.
print("\nLogarithm (torch.log(k)):\n", torch.log(k))

# Exponential
print("\nExponential (torch.exp(k)):\n", torch.exp(k))

# Square root
print("\nSquare Root (torch.sqrt(k)):\n", torch.sqrt(k))

# Sigmoid activation function
print("\nSigmoid (torch.sigmoid(k)):\n", torch.sigmoid(k))

# Softmax activation function (usually along a dimension)
print("\nSoftmax along dim=0 (torch.softmax(k, dim=0)):\n", torch.softmax(k, dim=0))

# Rectified Linear Unit (ReLU) activation function
print("\nReLU (torch.relu(k)):\n", torch.relu(k))
```

## 6. Inplace Operations



Inplace operations are operations that modify the tensor in place without creating a new one. They are typically denoted by a trailing underscore (e.g., `add_`, `relu_`). This can be memory efficient, but be cautious as it directly changes the original tensor.

```python
m = torch.rand(2,3)
n = torch.rand(2,3)

print("Original tensor m:\n", m)
print("Original tensor n:\n", n)

# Inplace addition: m is modified
m.add_(n)
print("\nm after m.add_(n):\n", m)
print("Tensor n (unchanged):\n", n)

# Non-inplace ReLU: returns a new tensor, m is unchanged
print("\nNon-inplace ReLU (torch.relu(m)):\n", torch.relu(m))
print("Tensor m (still original values):\n", m)

# Inplace ReLU: m is modified
m.relu_()
print("\nm after m.relu_():\n", m)
```

## 7. Copying a Tensor

Understanding how tensors are copied is crucial to avoid unintended side effects. Simply assigning one tensor to another (`b = a`) creates a new reference to the *same* underlying data, while `clone()` creates a new tensor with a copy of the data.

```python
a = torch.rand(2,3)
print("Original tensor a:\n", a)

# Case 1: Simple assignment (creates a view/reference)
b = a
print("\nTensor b (assigned from a):\n", b)

# Modifying 'a' will also modify 'b' because they point to the same memory
a[0][0] = 0
print("\na after modifying a[0][0]:\n", a)
print("b after modifying a[0][0]:\n", b)

# Check memory addresses to confirm they are the same object
print("\nID of a (memory address):", id(a))
print("ID of b (memory address):", id(b))

# Case 2: Cloning (creates a deep copy)
b = a.clone()
print("\na after cloning:\n", a)
print("b after cloning:\n", b)

# Now, modifying 'a' will NOT modify 'b'
a[0][0] = 10
print("\na after modifying a[0][0] again:\n", a)
print("b after modifying a[0][0] again (unchanged):\n", b)

# Check memory addresses to confirm they are different objects
print("\nID of a (memory address):", id(a))
print("ID of b (memory address):", id(b))
```
